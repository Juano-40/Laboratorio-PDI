```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('PDI P4.jpg', cv2.IMREAD_GRAYSCALE)
filas, columnas = img.shape
print("Dimensiones originales:", filas, "x", columnas)
```

**PARTE 1**
Ejercicio 1

```python
# Matriz de traslacion entera (50, 30)
M_entera = np.float32([[1, 0, 50], [0, 1, 30]])
img_tras_entera = cv2.warpAffine(img, M_entera, (columnas, filas))

# Matriz de traslacion decimal (20.5, 15.5)
M_decimal = np.float32([[1, 0, 20.5], [0, 1, 15.5]])
img_tras_decimal = cv2.warpAffine(img, M_decimal, (columnas, filas))

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(img, cmap='gray')
axes[0].set_title("Imagen Original")
axes[0].axis("off")

axes[1].imshow(img_tras_entera, cmap='gray')
axes[1].set_title("Traslacion (50, 30)")
axes[1].axis("off")

axes[2].imshow(img_tras_decimal, cmap='gray')
axes[2].set_title("Traslacion Decimal (20.5, 15.5)")
axes[2].axis("off")

plt.tight_layout()
plt.show()
```

Ejercicio 2

```python
centro = (columnas / 2, filas / 2)
M_rotacion = cv2.getRotationMatrix2D(centro, 45, 1.0)
img_rotada = cv2.warpAffine(img, M_rotacion, (columnas, filas))

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(img, cmap='gray')
axes[0].set_title("Imagen Original")
axes[0].axis("off")

axes[1].imshow(img_rotada, cmap='gray')
axes[1].set_title("Rotacion 45 Grados")
axes[1].axis("off")

plt.tight_layout()
plt.show()
```

Ejercicio 3


```python
img_150 = cv2.resize(img, None, fx=1.5, fy=1.5, interpolation=cv2.INTER_CUBIC)
img_50 = cv2.resize(img, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_AREA)

print("Tamaño 150%:", img_150.shape)
print("Tamaño 50%:", img_50.shape)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(img, cmap='gray')
axes[0].set_title(f"Original {img.shape}")

axes[1].imshow(img_150, cmap='gray')
axes[1].set_title(f"Escala 150% {img_150.shape}")

axes[2].imshow(img_50, cmap='gray')
axes[2].set_title(f"Escala 50% {img_50.shape}")

plt.tight_layout()
plt.show()
```

**Parte 2** Ejercicio 1

```python
img_bajo_contraste = cv2.imread('PDI P4 Parte 2.jpg', cv2.IMREAD_GRAYSCALE)

# Aplicar ecualizacion de histograma
img_ecualizada = cv2.equalizeHist(img_bajo_contraste)

# Graficar comparativa de imagenes y sus histogramas
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Imagen original
axes[0, 0].imshow(img_bajo_contraste, cmap='gray', vmin=0, vmax=255)
axes[0, 0].set_title("Imagen Medica Original (Bajo Contraste)")
axes[0, 0].axis("off")

# Histograma original
axes[0, 1].hist(img_bajo_contraste.ravel(), bins=256, range=[0, 256], color='black')
axes[0, 1].set_title("Histograma Original")
axes[0, 1].set_xlabel("Nivel de Intensidad")
axes[0, 1].set_ylabel("Cantidad de Pixeles")
axes[0, 1].set_xlim([0, 256])

# Imagen ecualizada
axes[1, 0].imshow(img_ecualizada, cmap='gray', vmin=0, vmax=255)
axes[1, 0].set_title("Imagen Ecualizada (Contraste Mejorado)")
axes[1, 0].axis("off")

# Histograma ecualizado
axes[1, 1].hist(img_ecualizada.ravel(), bins=256, range=[0, 256], color='black')
axes[1, 1].set_title("Histograma Ecualizado")
axes[1, 1].set_xlabel("Nivel de Intensidad")
axes[1, 1].set_ylabel("Cantidad de Pixeles")
axes[1, 1].set_xlim([0, 256])

plt.tight_layout()
plt.show()
```
